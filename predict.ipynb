```python
def assign_priority(ticket):
    ticket = ticket.lower()

    critical_words = [
        "down",
        "outage",
        "completely unavailable",
        "all users",
        "production down",
        "system unavailable"
    ]

    high_words = [
        "urgent",
        "critical",
        "cannot login",
        "not working",
        "blocked",
        "failure",
        "unable to access"
    ]

    medium_words = [
        "issue",
        "problem",
        "error",
        "request",
        "need help"
    ]

    if any(word in ticket for word in critical_words):
        return "Critical"

    elif any(word in ticket for word in high_words):
        return "High"

    elif any(word in ticket for word in medium_words):
        return "Medium"

    else:
        return "Low"
```
